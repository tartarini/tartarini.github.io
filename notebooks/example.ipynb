## Generate noisy data and fit a line

```python
rng = np.random.default_rng(42)
x = np.linspace(0, 10, 50)
y = 2.5 * x + 1.0 + rng.normal(0, 2, x.size)

slope, intercept = np.polyfit(x, y, 1)
print(f"Fitted: y = {slope:.2f} x + {intercept:.2f}")
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(x, y, s=18, label="data")
plt.plot(x, slope * x + intercept, color="C1", label="fit")
plt.xlabel("x"); plt.ylabel("y"); plt.legend(); plt.tight_layout()
plt.show()
```
